```python
#Write a program to check if a given character is a vowel or consonant.

vowels = ["a","e","i","o","u"]

ch=input("enter ch:")
if(ch in vowels):
    print(f"{ch} is a vowel")
else:
    print(f"{ch} is a consonant")
```

```python
#Write a program to print the sum of digits of a number.

a=int(input("enter a:"))

sum=0
temp=a
while(temp//10 > 0):
    c=temp%10
    sum+=c
    temp=temp//10
sum+=temp

print(f"sum of 3 digit= {sum} ")
```

```python
#Write a program to count how many digits are present in a number.

a=int(input("enter a:"))

count=0
temp=a
while(temp//10 > 0):
    c=temp%10
    count+=1
    temp=temp//10
count+=1

print(f"count of 3 digit= {count} ")
```

```python
#Write a program to find the factorial of a number.

a=int(input("enter a:"))
fact=1
for i in range (1,a+1):
    fact*=i

print(f"fact of {a} = {fact}")
```

```python
#Write a program to print all prime numbers from 1 to 50.

import math

def is_prime(n):
    count=0
    for i in range (2,int(math.sqrt(n)+1)):
        if(n%i == 0):
            count+=1

    if(count<1):
        return True
    else:
        return False

for i in range (1,51):
    if(is_prime(i)):
        print(f"{i} is prime")
```
